# ポートフォリオ分析の基礎

このNotebookでは、期待リターン、共分散行列、ポートフォリオのリスク、ランダムポートフォリオによる効率的フロンティアの近似を確認します。

ここで描く効率的フロンティアは、最適化ではなくランダムに生成したロングオンリーの重みを可視化した教育用の近似です。

```python
import numpy as np

from finsimlab.portfolio import (
    generate_random_portfolios,
    plot_efficient_frontier,
    portfolio_return,
    portfolio_volatility,
    rebalance_portfolio,
)
```

## リターンと共分散行列（分散投資の「種明かし」）

ポートフォリオ運用の基本は、**「卵を一つのカゴに盛るな」**という格言に集約されます。一つのカゴ（資産）を落としても、他のカゴが 無事なら致命傷を避けられるという考え方です。

### 例え：アイスクリーム屋さんと傘屋さん
- **相関が高い**：晴れの日に両方売れる「アイスクリーム屋」と「ジュース屋」。暑ければ儲かりますが、涼しいと両方ダメになります。
- **相関が低い（または負）**：「アイスクリーム屋」と「傘屋」。晴れればアイスが売れ、雨が降れば傘が売れるため、天候に関わらず売上が安定します。

共分散行列は、こうした「資産同士の相性の良し悪し（連動性）」を数値化したものです。

```python
expected_returns = np.array([0.04, 0.07, 0.10])
covariance_matrix = np.array([
    [0.0100, 0.0018, 0.0011],
    [0.0018, 0.0225, 0.0026],
    [0.0011, 0.0026, 0.0400],
])

expected_returns, covariance_matrix
```

## 1つのポートフォリオを評価する

```python
weights = np.array([0.5, 0.3, 0.2])

portfolio_return(expected_returns, weights), portfolio_volatility(covariance_matrix, weights)
```

## ランダムポートフォリオで可視化する

```python
portfolios = generate_random_portfolios(
    expected_returns,
    covariance_matrix,
    n_portfolios=3000,
    risk_free_rate=0.01,
    seed=42,
)

plot_efficient_frontier(portfolios)
```

## リバランスの直感（庭の手入れ）

ポートフォリオを放置すると、値上がりした資産の割合が大きくなり、当初の予定よりリスクを取りすぎてしまうことがあります。

### 例え：庭木の剪定（せんてい）
リバランスは、伸びすぎた枝（値上がりした資産）を切り、足りない部分（値下がりした資産）に栄養を回す**「庭の手入れ」**のようなものです。定期的に行うことで、ポートフォリオを常に自分にとって心地よい状態（リスク水準）に保つことができます。

```python
rebalance_portfolio([3000, 5000, 2000], [0.4, 0.4, 0.2])
```

## 学習メモ: 直感・数式・演習

### 直感

ポートフォリオ分析では、個別資産を単独で見るのではなく、組み合わせたときのリスクとリターンを見ます。重要なのは、リスクは単純な足し算ではないという点です。資産同士の相関が低ければ、一方が下がったときにもう一方が支える場合があり、全体のぶれが小さくなることがあります。

このNotebookの効率的フロンティアは、厳密な最適化ではなく、ランダムに作った多数のウェイトを散布図で見ている近似的な可視化です。初心者が「重みを変えるとリスク・リターンがどう動くか」をつかむためのものです。

### 数式の最小説明

ポートフォリオの期待リターンは、各資産の期待リターンをウェイトで加重平均します。

$$E[R_p] = w^T \mu$$

分散は、ウェイトと共分散行列を使って計算します。

$$\sigma_p^2 = w^T \Sigma w$$

ここで、$w$ はウェイト、$\mu$ は期待リターン、$\Sigma$ は共分散行列です。ボラティリティは分散の平方根です。
