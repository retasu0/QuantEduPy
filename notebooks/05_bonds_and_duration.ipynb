## 学習メモ: 直感・数式・演習

### 直感

債券価格は、将来受け取るクーポンと額面を現在価値に割り引いた合計です。利回りが上がると、将来キャッシュフローの現在価値は小さくなるため、固定利付債の価格は下がります。

デュレーションは、債券価格が利回り変化にどれくらい敏感かを知るための基本指標です。Macaulay durationはキャッシュフローを受け取るまでの加重平均期間として、modified durationは利回り変化に対する価格感応度として理解できます。Convexityは、利回り変化が大きいときに線形近似だけでは足りない部分を補います。

### 数式の最小説明

現在価値は、将来キャッシュフローを利率で割り引いて合計します。

$$PV = \sum_{t} \frac{CF_t}{(1+r)^t}$$

Macaulay durationは、現在価値で重み付けした受取時点の平均です。

$$D_M = \frac{\sum_t t \cdot PV(CF_t)}{\sum_t PV(CF_t)}$$

Modified durationは、Macaulay durationを利回りの複利頻度で調整します。

$$D_{mod} = \frac{D_M}{1 + y/m}$$

ここで、$y$ は利回り、$m$ は年あたりの支払回数です。
